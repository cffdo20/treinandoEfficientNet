```python
import os
import pandas as pd

def process_csv(input_csv, output_folder):
    # Lê o CSV
    df = pd.read_csv(input_csv)

    # Verifica se a coluna 'layers' existe, e alterna os valores entre -135 e -100
    if 'layers' in df.columns:
        df['layers'] = [-100 if i % 2 != 0 else -135 for i in range(len(df))]

    # Remove a coluna 'history', se existir
    if 'history' in df.columns:
        df.drop(columns=['history'], inplace=True)

    # Gera o caminho do arquivo de saída
    output_csv = os.path.join(output_folder, 'result_treino10_modificado.csv')

    # Salva o novo CSV
    df.to_csv(output_csv, index=False)

    print(f"Arquivo modificado salvo em: {output_csv}")

```

```python
# Exemplo de uso
input_csv = 'C:/Users/PC/Documents/GitHub/treinandoEficcientNet2/resultados/results_treinoLote10_.csv'  # Caminho do arquivo de entrada
output_folder = 'C:/Users/PC/Documents/GitHub/treinandoEficcientNet2/resultados'  # Pasta de destino
process_csv(input_csv, output_folder)
```
